```python
required_columns = [
    "Ticket Subject",
    "Ticket Description",
    "Ticket Type",
    "Ticket Priority"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

data = df[required_columns].copy()

# Remove rows missing any input or target
data = data.dropna(subset=required_columns)

# Convert values to strings and remove surrounding whitespace
for col in required_columns:
    data[col] = data[col].astype(str).str.strip()

# Remove rows with empty text or empty target labels
data = data[
    (data["Ticket Subject"] != "") &
    (data["Ticket Description"] != "") &
    (data["Ticket Type"] != "") &
    (data["Ticket Priority"] != "")
].copy()

# Combine subject and description
data["Ticket Text"] = (
    data["Ticket Subject"] + " " + data["Ticket Description"]
)

# Remove exact duplicate examples to reduce repeated-example bias
data = data.drop_duplicates(
    subset=["Ticket Text", "Ticket Type", "Ticket Priority"]
).reset_index(drop=True)

print("Rows available for training:", len(data))
print("\nCategory labels:", sorted(data["Ticket Type"].unique()))
print("\nPriority labels:", sorted(data["Ticket Priority"].unique()))

display(data[["Ticket Text", "Ticket Type", "Ticket Priority"]].head())
```

```python
def clean_text(text):
    text = str(text).lower()

    # Remove placeholders such as {product_purchased}
    text = re.sub(r"\{[^}]*\}", " ", text)

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Keep letters, numbers, and whitespace
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # Remove repeated spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text

data["Ticket Text"] = data["Ticket Text"].apply(clean_text)

# Remove rows where cleaning leaves no usable text
data = data[data["Ticket Text"].str.len() > 0].reset_index(drop=True)

print("Text cleaning completed.")
print("Final usable rows:", len(data))
print("\nExample cleaned ticket:")
print(data["Ticket Text"].iloc[0])
```

```python
# Check that every target class has enough examples for a stratified split
for target in ["Ticket Type", "Ticket Priority"]:
    counts = data[target].value_counts()
    rare = counts[counts < 2]

    if not rare.empty:
        print(f"\nWarning: rare labels in {target}:")
        print(rare)

print("\nData preparation checks complete.")
```

```python
def build_model(algorithm):
    if algorithm == "Logistic Regression":
        classifier = LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        )

    elif algorithm == "Multinomial Naive Bayes":
        classifier = MultinomialNB()

    elif algorithm == "Linear SVM":
        classifier = LinearSVC(
            class_weight="balanced",
            random_state=42
        )

    else:
        raise ValueError("Unknown algorithm")

    return Pipeline([
        (
            "tfidf",
            TfidfVectorizer(
                ngram_range=(1, 2),
                max_features=30000,
                min_df=2,
                sublinear_tf=True
            )
        ),
        ("classifier", classifier)
    ])


print("Model-building function created successfully.")
```

```python
algorithms = [
    "Logistic Regression",
    "Multinomial Naive Bayes",
    "Linear SVM"
]

results = []
trained_models = {}
test_data = {}

def train_and_evaluate(target):
    print("\n" + "=" * 65)
    print("PREDICTION TASK:", target)
    print("=" * 65)

    X = data["Ticket Text"]
    y = data[target]

    # Stratify when every class has enough examples
    stratify_labels = y if y.value_counts().min() >= 2 else None

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=stratify_labels
    )

    test_data[target] = (X_test, y_test)

    best_name = None
    best_score = -1

    for algorithm in algorithms:
        model = build_model(algorithm)
        model.fit(X_train, y_train)

        predictions = model.predict(X_test)

        accuracy = accuracy_score(y_test, predictions)
        precision = precision_score(
            y_test, predictions, average="macro", zero_division=0
        )
        recall = recall_score(
            y_test, predictions, average="macro", zero_division=0
        )
        macro_f1 = f1_score(
            y_test, predictions, average="macro", zero_division=0
        )

        results.append({
            "Target": target,
            "Algorithm": algorithm,
            "Accuracy": accuracy,
            "Macro Precision": precision,
            "Macro Recall": recall,
            "Macro F1": macro_f1
        })

        trained_models[(target, algorithm)] = model

        print(f"\n--- {algorithm} ---")
        print(f"Accuracy: {accuracy:.4f}")
        print(f"Macro F1: {macro_f1:.4f}")

        print("\nClassification report:")
        print(classification_report(
            y_test,
            predictions,
            zero_division=0
        ))

        if macro_f1 > best_score:
            best_score = macro_f1
            best_name = algorithm

    print(f"\nBest algorithm for {target}: {best_name}")
    print(f"Best macro F1-score: {best_score:.4f}")

    return best_name


best_category_algorithm = train_and_evaluate("Ticket Type")
best_priority_algorithm = train_and_evaluate("Ticket Priority")

print("\nBoth training tasks completed!")
```

```python

from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import joblib
from pathlib import Path

clean_results = []
final_models = {}
final_data = {}

for target in ["Ticket Type", "Ticket Priority"]:
    print(f"\n===== {target} =====")

    # Keep only texts that have one consistent label
    label_counts = data.groupby("Ticket Text")[target].nunique()
    valid_texts = label_counts[label_counts == 1].index

    clean_data = (
        data[data["Ticket Text"].isin(valid_texts)]
        .drop_duplicates(subset=["Ticket Text"])
        .copy()
    )

    X = clean_data["Ticket Text"]
    y = clean_data[target]

    print("Usable unique texts:", len(clean_data))
    print("Excluded ambiguous texts:", len(data["Ticket Text"].unique()) - len(valid_texts))

    # Split unique texts so identical text cannot leak across sets
    stratify_labels = y if y.value_counts().min() >= 2 else None

    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42,
        stratify=stratify_labels
    )

    target_models = [
        (algorithm, model)
        for (model_target, algorithm), model in trained_models.items()
        if model_target == target
    ]

    if not target_models:
        raise ValueError(f"No previously trained models found for {target}.")

    scores = []

    # Retrain and evaluate every previously tested algorithm
    for algorithm, original_model in target_models:
        model = clone(original_model)
        model.fit(X_train, y_train)
        predictions = model.predict(X_test)

        macro_f1 = f1_score(
            y_test, predictions, average="macro", zero_division=0
        )

        scores.append({
            "Target": target,
            "Algorithm": algorithm,
            "Accuracy": accuracy_score(y_test, predictions),
            "Macro Precision": precision_score(
                y_test, predictions, average="macro", zero_division=0
            ),
            "Macro Recall": recall_score(
                y_test, predictions, average="macro", zero_division=0
            ),
            "Macro F1": macro_f1,
            "Model": model
        })

    scores.sort(key=lambda item: item["Macro F1"], reverse=True)
    best = scores[0]

    print("Best algorithm:", best["Algorithm"])
    print("Accuracy:", round(best["Accuracy"], 4))
    print("Macro F1:", round(best["Macro F1"], 4))

    # Retrain the best algorithm on all clean, unambiguous examples
    final_model = clone(dict(target_models)[best["Algorithm"]])
    final_model.fit(X, y)

    final_models[target] = final_model
    final_data[target] = clean_data

    for result in scores:
        clean_results.append({
            key: value
            for key, value in result.items()
            if key != "Model"
        })

# Save the new models and comparison report
MODELS_DIR = Path.cwd() / "models"
OUTPUTS_DIR = Path.cwd() / "outputs"
MODELS_DIR.mkdir(parents=True, exist_ok=True)
OUTPUTS_DIR.mkdir(parents=True, exist_ok=True)

joblib.dump(final_models["Ticket Type"], MODELS_DIR / "ticket_type_model.joblib")
joblib.dump(final_models["Ticket Priority"], MODELS_DIR / "ticket_priority_model.joblib")

clean_results_df = __import__("pandas").DataFrame(clean_results)
clean_results_df.to_csv(
    OUTPUTS_DIR / "clean_model_comparison.csv", index=False
)

print("\nStep 2A completed.")
print("Updated models saved in:", MODELS_DIR.resolve())
print("Clean comparison saved in:", OUTPUTS_DIR.resolve())
```
